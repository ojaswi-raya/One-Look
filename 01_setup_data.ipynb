# One Look — Stage 1: download Sen1Floods11

This notebook downloads the hand-labelled Sen1Floods11 data to Google Drive. It can take time and requires sufficient Drive storage. Do not share the downloaded dataset publicly in this GitHub repository.

```python
from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
DATA = Path('/content/drive/MyDrive/one-look/data/HandLabeled')
DATA.mkdir(parents=True, exist_ok=True)
print('Dataset destination:', DATA)
```

## Download the three required folders
`LabelHand` contains water masks, `S2Hand` contains Sentinel-2 multispectral images, and `JRCWaterHand` identifies permanent water. This download may take a while.

```python
import subprocess
for folder in ['LabelHand', 'S2Hand', 'JRCWaterHand']:
    target = DATA / folder
    if target.exists() and any(target.iterdir()):
        print(f'{folder}: already present; skipping download')
    else:
        print(f'Downloading {folder}...')
        cmd = ['gsutil', '-m', 'cp', '-r', f'gs://sen1floods11/v1.1/data/flood_events/HandLabeled/{folder}', str(DATA) + '/']
        subprocess.run(cmd, check=True)
```
